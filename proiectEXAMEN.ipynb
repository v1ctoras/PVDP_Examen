# Introducere & Ipoteză de lucru & Descriere scurtă a setului de date

Testul <font color = "red">PISA</font> apărut în anul 1997, se mai spune și studiul internațional al studenților organizat de OCDE ↔ Organizația pentru Cooperare & Dezvoltare Economică, este orientat pe <font color = "red">evaluarea compentențelor de bază ale elevilor de ~15 ani</font>.

Ideea generală este că nu este vorba despre cât de bine au <font color = "red">memorat elevii</font> materiile școlare, ci capacitatea de a <font color = "red">aplica cunoștințele acumulate </font> care pot fi folosiți pentru rezolvarea problemelor ce pot apărea în viață → provocări neprevăzute în societate. De exemplu, putem lua în considerare testarea coeficientului de inteligență [IQ].

Scopul testelor PISA este ca fiecare țară să-și <font color = "red">imbunătățească mediul educațional</font> → decizii guvernamentale în privința investițiilor & perfomanțele sistemelor de educație prin evaluări periodice. Din punct de vedere, impactul evaluărilor internaționale este concludent. Adevărul este că când vine vorba mereu despre 'educație', influența politică a existat în mod direct & indirect și există întotdeauna.

Acest proiect construit are următoarele 3 ipoteze principale:

```python
# construire reprezentare grafica prin scatter-plot
import numpy as np
import matplotlib.pyplot as plt

# construire variabile care au valori din fiecare țară
data_fete_m = set_de_date.iloc[1:, set_de_date.columns.get_loc('MathGirl')]
data_baieti_m = set_de_date.iloc[1:, set_de_date.columns.get_loc('MathBoy')]

data_fete_r = set_de_date.iloc[1:, set_de_date.columns.get_loc('ReadGirl')]
data_baieti_r = set_de_date.iloc[1:, set_de_date.columns.get_loc('ReadBoy')]

data_fete_s = set_de_date.iloc[1:, set_de_date.columns.get_loc('ScienceGirl')]
data_baieti_s = set_de_date.iloc[1:, set_de_date.columns.get_loc('ScienceBoy')]

# construire variabile care au valori din media OECD
data_OECD_fete_m = set_de_date.iloc[0, set_de_date.columns.get_loc('MathGirl')]
data_OECD_baieti_m = set_de_date.iloc[0, set_de_date.columns.get_loc('MathBoy')]

data_OECD_fete_r = set_de_date.iloc[0, set_de_date.columns.get_loc('ReadGirl')]
data_OECD_baieti_r = set_de_date.iloc[0, set_de_date.columns.get_loc('ReadBoy')]

data_OECD_fete_s = set_de_date.iloc[0, set_de_date.columns.get_loc('ScienceGirl')]
data_OECD_baieti_s = set_de_date.iloc[0, set_de_date.columns.get_loc('ScienceBoy')]

# afișarea graficelor în 3 moduri
fix, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize = (12,10))

# construire grafice de tip scatter plot - mod I [matematica] & adăugare medie
ax1.scatter(range(len(data_fete_m)), data_fete_m, color = 'pink', label = 'scor fete', s=25)
ax1.scatter(range(len(data_baieti_m)), data_baieti_m, color = 'blue', label = 'scor băieți', s=25)
ax1.axhline(y=data_OECD_fete_m, color = 'pink', linestyle = '--', linewidth = 2, label = 'medie OECD')
ax1.axhline(y=data_OECD_baieti_m, color = 'blue', linestyle = '--', linewidth = 2, label = 'medie OECD')
ax1.set_xticks([])
ax1.set_ylabel('Scor PISA')
ax1.set_title('Reprezentarea grafică [scatter-plot] a rezultatelor obținute privind compentența de matematică F vs B')

ax1.legend()

# construire grafice de tip scatter plot - mod II [lectură] & adăugare medie
ax2.scatter(range(len(data_fete_r)), data_fete_r, color = 'pink', label = 'fete', s=25)
ax2.scatter(range(len(data_baieti_r)), data_baieti_r, color = 'blue', label = 'băieți', s=25)
ax2.axhline(y=data_OECD_fete_r, color = 'pink', linestyle = '--', linewidth = 2, label = 'medie OECD fete')
ax2.axhline(y=data_OECD_baieti_r, color = 'blue', linestyle = '--', linewidth = 2, label = 'medie OECD băieți')
ax2.set_xticks([])
ax2.set_ylabel('Scor PISA')
ax2.set_title('Reprezentarea grafică [scatter-plot] a rezultatelor obținute privind compentența de lectură F vs B')

# construire grafice de tip scatter plot - mod III [știință] & adăugare medie
ax3.scatter(range(len(data_fete_s)), data_fete_s, color = 'pink', label = 'fete', s=25)
ax3.scatter(range(len(data_baieti_s)), data_baieti_s, color = 'blue', label = 'băieți', s=25)
ax3.axhline(y=data_OECD_fete_s, color = 'pink', linestyle = '-', linewidth = 2, label = 'medie OECD fete')
ax3.axhline(y=data_OECD_baieti_s, color = 'blue', linestyle = 'dotted', linewidth = 2, label = 'medie OECD băieți')
ax3.set_xticks([])
ax3.set_ylabel('Scor PISA')
ax3.set_title('Reprezentarea grafică [scatter-plot] a rezultatelor obținute privind compentența de știință F vs B')

plt.tight_layout()
plt.show()
```

```python
# construire reprezentare grafica de tip triplă-bare care conțin mai multe valori comparative
import numpy as np
import matplotlib.pyplot as plt

# construire variabilă pentru fiecare țară în domeniul matematic
bar1_RO = set_de_date.iloc[64, set_de_date.columns.get_loc('Math')]
bar1_MD = set_de_date.iloc[48, set_de_date.columns.get_loc('Math')]
bar1_BG = set_de_date.iloc[9, set_de_date.columns.get_loc('Math')]
bar1_UKR = set_de_date.iloc[75, set_de_date.columns.get_loc('Math')]
bar1_HR = set_de_date.iloc[30, set_de_date.columns.get_loc('Math')]
bar1_SR = set_de_date.iloc[66, set_de_date.columns.get_loc('Math')]
bar1_OECD = set_de_date.iloc[0, set_de_date.columns.get_loc('Math')]

# construire variabilă pentru fiecare țară în domeniul lecturii
bar2_RO = set_de_date.iloc[64, set_de_date.columns.get_loc('Read')]
bar2_MD = set_de_date.iloc[48, set_de_date.columns.get_loc('Read')]
bar2_BG = set_de_date.iloc[9, set_de_date.columns.get_loc('Read')]
bar2_UKR = set_de_date.iloc[75, set_de_date.columns.get_loc('Read')]
bar2_HR = set_de_date.iloc[30, set_de_date.columns.get_loc('Read')]
bar2_SR = set_de_date.iloc[66, set_de_date.columns.get_loc('Read')]
bar2_OECD = set_de_date.iloc[0, set_de_date.columns.get_loc('Read')]

# construire variabilă pentru fiecare țară în domeniul științific
bar3_RO = set_de_date.iloc[64, set_de_date.columns.get_loc('Science')]
bar3_MD = set_de_date.iloc[48, set_de_date.columns.get_loc('Science')]
bar3_BG = set_de_date.iloc[9, set_de_date.columns.get_loc('Science')]
bar3_UKR = set_de_date.iloc[75, set_de_date.columns.get_loc('Science')]
bar3_HR = set_de_date.iloc[30, set_de_date.columns.get_loc('Science')]
bar3_SR = set_de_date.iloc[66, set_de_date.columns.get_loc('Science')]
bar3_OECD = set_de_date.iloc[0, set_de_date.columns.get_loc('Science')]

# adăugare totală în cele 3 bare
bar1 = [bar1_RO, bar1_MD, bar1_BG, bar1_UKR, bar1_HR, bar1_SR, bar1_OECD]
bar2 = [bar2_RO, bar2_MD, bar2_BG, bar2_UKR, bar2_HR, bar2_SR, bar2_OECD]
bar3 = [bar3_RO, bar3_MD, bar3_BG, bar3_UKR, bar3_HR, bar3_SR, bar3_OECD]

x = np.arange(7)
width = 0.15

fig, ax = plt.subplots(figsize = (12, 8))

# afișare triplă-bare unde sunt setate dimensiunile & culori
a = ax.bar(x-width, bar1, width, label = 'scor PISA matematică', color = '#8A9A5B')
b = ax.bar(x, bar2, width, label = 'scor PISA lectură', color = '#6082B6')
c = ax.bar(x+width, bar3, width, label = 'scor PISA știință', color = '#CF9FFF')

# afișare valori pentru fiecare bară
ax.bar_label(a, padding = 1, fmt = '%.0f', fontsize = 9, color = '#8A9A5B')
ax.bar_label(b, padding = 1, fmt = '%.0f', fontsize = 9, color = '#6082B6')
ax.bar_label(c, padding = 1, fmt = '%.0f', fontsize = 9, color = '#CF9FFF')

# afișare medie sub formă de linie
plt.axhline(y = bar1_OECD, color = '#8A9A5B', linestyle = 'dotted', linewidth = 2)
plt.axhline(y = bar2_OECD, color = '#6082B6', linestyle = 'dotted', linewidth = 2)
plt.axhline(y = bar3_OECD, color = '#CF9FFF', linestyle = 'dotted', linewidth = 2)

ax.legend()
ax.set_title('Perfomanța României & țările vecine la testele PISA 2022')
ax.set_xticks([0, 1, 2, 3, 4, 5, 6])
ax.set_xticklabels(['România', 'Moldova', 'Bulgaria', 'Ucraina', 'Ungaria', 'Serbia', 'Medie OECD'])

# comandă pentru afișarea detaliată
ax.set_ylim(300, 550)

plt.tight_layout()
plt.show()
```
