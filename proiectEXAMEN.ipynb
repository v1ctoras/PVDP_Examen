# Introducere & Ipoteză de lucru & Descriere scurtă a setului de date

Testul <font color = "red">PISA</font> apărut în anul 1997, se mai spune și studiul internațional al studenților organizat de OCDE ↔ Organizația pentru Cooperare & Dezvoltare Economică, este orientat pe <font color = "red">evaluarea compentențelor de bază ale elevilor de ~15 ani</font>.

Ideea generală este că nu este vorba despre cât de bine au <font color = "red">memorat elevii</font> materiile școlare, ci capacitatea de a <font color = "red">aplica cunoștințele acumulate </font> care pot fi folosiți pentru rezolvarea problemelor ce pot apărea în viață → provocări neprevăzute în societate. De exemplu, putem lua în considerare testarea coeficientului de inteligență [IQ].

Scopul testelor PISA este ca fiecare țară să-și <font color = "red">imbunătățească mediul educațional</font> → decizii guvernamentale în privința investițiilor & perfomanțele sistemelor de educație prin evaluări periodice. Din punct de vedere, impactul evaluărilor internaționale este concludent. Adevărul este că când vine vorba mereu despre 'educație', influența politică a existat în mod direct & indirect și există întotdeauna.

Acest proiect construit are următoarele 3 ipoteze principale:


*   Media rezultatelor obținute din testele PISA pentru fiecare continent [Asia, Africa, America de Nord, America de Sud, Antarctica, Europa & Australia]. Pe baza rezultatelor statistice, care continent are elevi cu rezultatele mai bune în cadrul testelor PISA 2022?
*   Așa cum susține lumea, este adevărat că fetele au perfomanțe mai bune decât băieții?
*   Care este poziția României & țările vecine [Moldova, Bulgaria, Ucraina, Ungaria & Serbia] față de media OECD?

De regulă, testarea PISA se concentrează pe 3 domenii fundamentale:

*   `Math` → matematică → înțelegerea & rezolvarea problemelor practice
*   `Read` → lectură → înțelegerea & identificarea informațiilor necesare
*   `Science` → știință → înțelegerea & analizarea datelor

Pe lângă acest lucru, se mai adaug niște variabile referitoare la:
